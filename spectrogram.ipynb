# Usage
This notebook use for convert Audio folder to spectrogram

```python
import os

from torch.utils.data import Dataset
import pandas as pd
import torchaudio

import matplotlib.pyplot as plt
import numpy as np
```

```python
SAMPLE_RATE = 16000
```

```python
mel_spectrogram = torchaudio.transforms.MelSpectrogram(
    sample_rate= SAMPLE_RATE,
    n_fft= 400,
    hop_length=160,
    win_length=400,
    n_mels=80,
    center=False
)
```

```python
AUDIO_FOLDER_PATH = "C:/Users/Heathcliff/Desktop/SR from scratch/FOSD Dataset processing/resampled_audio"

SPECTROGRAM_PATH = "Spectrogram"
if not os.path.isdir(SPECTROGRAM_PATH):
    os.makedirs(SPECTROGRAM_PATH)
```

```python
import glob
```

```python
audio_files = glob.glob(os.path.join(AUDIO_FOLDER_PATH, '*.mp3'))
# count = 1

for file in audio_files:
    ## loop for continue when interupt
    # if count <= 16139:
    #     count +=1 
    #     continue
    waveform, sr = torchaudio.load(file)

    mel_spec = mel_spectrogram(waveform)

    mel_spectrogram_db = torchaudio.transforms.AmplitudeToDB()(mel_spec)

    # Chọn kênh đầu tiên (nếu âm thanh có nhiều kênh)
    mel_spectrogram_db = mel_spectrogram_db[0].numpy()

    # Vẽ Mel Spectrogram
    plt.figure(figsize=(10, 4))
    plt.imshow(mel_spectrogram_db, aspect='auto', origin='lower', cmap='inferno')
    plt.title('Mel Spectrogram')
    plt.xlabel('Time')
    plt.ylabel('Frequency')
    plt.colorbar(format='%+2.0f dB')

    # Lưu hình ảnh
    spectrogram_path = os.path.join(SPECTROGRAM_PATH, os.path.basename(file).replace(".mp3", ".png"))
    plt.savefig(spectrogram_path, bbox_inches='tight')
    plt.close()
```
